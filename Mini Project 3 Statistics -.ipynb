```python
# Do Males and Females spend differently
```

```python
#Finding male spendings
male_spending = df[df['Gender'] == 'Male']['MonthlySpend']
print('Male spendings are as follows :\n',male_spending,'\n')
female_spending = df[df['Gender'] == 'Female']['MonthlySpend']
print('Female spendings are as follows :\n',female_spending)
```

```python

#gms=df.groupby('Gender')['MonthlySpend'].count()
#gms
```

```python


# We will find out the Male and Female spending behaviour using the Independent Two-Sample T-Test
# Considering equal_var = False as it ususally is for practical purposes.

t_stat, p_value = stats.ttest_ind(male_spending, female_spending, equal_var=False)

# 4. Output the results
alpha = 0.05
print(f"T-Statistic: {t_stat:.4f}")
print(f"P-Value: {p_value:.6f}")

if p_value < alpha:
    print("Result: Reject null hypothesis. There is a SIGNIFICANT difference in spending between males and females.")
else:
    print("Result: Fail to reject null hypothesis. There is NO significant difference in spending between males and females.")
```

```python
# Does education level impact average monthlyspend?
mse=df.groupby('Education')['MonthlySpend'].mean()
mse
```

```python
from scipy.stats import f_oneway

# Grouping the data and extracting MonthlySpend for each group
grouped_data = [group['MonthlySpend'].values for name, group in df.groupby('Education')]

# Unpacking the groups into f_oneway
f_stat, p_val = stats.f_oneway(*grouped_data)

print(f"F-Statistic: {f_stat:.4f}")
print(f"P-Value: {p_val:.6f}")


if p_val < 0.05:
    print("\nReject the null hypothesis: There is a significant difference between the group means.")
else:
    print("\nFail to reject the null hypothesis: No significant difference found.\n\n The F-statistic of",f_stat,
           'is < 1.0 which tells that the difference between the Monthly Spending amongst groups based on Education levels are much smaller than variation within the groups. \nThe P-Value :', p_val, 
        'Indicates that its a 92% chance that any small difference between groups is mostly random noise.'
        '\nConclusion : Education does not affect Monthly Spend in the given case.')
```

```python
#Calculating Confidence Interval
from scipy.stats import t

diff = married.mean() - unmarried.mean()

# To find standard error
se = np.sqrt(
    married.var(ddof=1)/len(married) +      
    unmarried.var(ddof=1)/len(unmarried)
)

dfree = len(married)+len(unmarried)-2

margin = t.ppf(0.975, dfree)*se  #Calculates the 95% confidence margin around the estimated difference
print('95% confidence margin :',margin)
lower = diff - margin
upper = diff + margin

print('Lower Bound :',lower,'\nUpper Bound :', upper)
print('\n\nWe are 95% confident that the true spending difference between married and unmarried customers lies between -$7.10 and +$9.9')

```

```python
#Calculating confidence level

n = len(df)

# Fisher Z transformation. Transforms the correlation into a scale where confidence intervals 
                                   # can be calculated accurately
z = np.arctanh(corr)

# Standard error #Measures uncertainty in the correlation estimate.
se = 1 / np.sqrt(n - 3)

# 95% CI in z-space
z_lower = z - 1.96 * se
z_upper = z + 1.96 * se

# Convert back to correlation scale (-1 to 1)
r_lower = np.tanh(z_lower)
r_upper = np.tanh(z_upper)

print("Correlation:", corr)
print("95% CI-",'\nLower Bound :',r_lower,'\nUpper Bound :', r_upper)
print('\n\nTherefore, we are 95% confident that the true correlation between engagement and spending lies between',r_lower,'and',r_upper)
print('\nAlso,The confidence interval is centered very close to zero and contains only extremely small values.This suggests :',
      '\nEven if a relationship exists between engagement and spending, it is too small to be practically meaningful.')
```
